## Part 1 — Classification with `MLPClassifier` (Wine Dataset)

The Wine dataset is a multi-class classification problem with 13 numeric features and 3 classes. We'll build MLPs inside a `Pipeline` with `StandardScaler`.

### 1.1 Load & split the data
- Use `train_test_split` with `test_size=0.2`, `stratify=y`, and `random_state=27`.


```python
from sklearn.datasets import load_wine, load_diabetes
from sklearn.model_selection import train_test_split, GridSearchCV
# TODO: add other import statements as needed
```

```python
# 1.1 Load & split
wine = load_wine()
X, y = wine.data, wine.target

# TODO: use train_test_split to separate the training and testing data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=27, stratify=y)
# print("X_train:", X_train.shape)
# print("X_test:", X_test.shape)
# print("y_train:", y_train.shape)
# print("y_test:", y_test.shape)
```


### 1.2 Baseline
Train a simple **Logistic Regression** baseline and report the performance of the model using appropriate metrics. Remember to use `StandardScaler` before training the model in the `Pipeline`.

```python
# TODO: build a Pipeline using StandardScaler and LogisticRegression.
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score
pipe = make_pipeline(
    StandardScaler(),
    LogisticRegression()
)
pipe.fit(X_train, y_train)
accuracy = pipe.score(X_test, y_test)
print(f"Test Accuracy: {accuracy:.2f}")
precision = precision_score(y_test, pipe.predict(X_test), average='weighted')
print(f"Test Precision: {precision:.2f}")
recall = recall_score(y_test, pipe.predict(X_test), average='weighted')
print(f"Test Recall: {recall:.2f}")

# print("Training samples:", X_train.shape[0])
# print("Testing samples:", X_test.shape[0])
# print("Actual classes:", np.unique(y_test))
# y_pred = pipe.predict(X_test)
# print("Predicted classes:", np.unique(y_pred))
# print("Prediction counts:", np.bincount(y_pred))
```


### 1.3 Build an MLP pipeline
Create a pipeline with `StandardScaler` and `MLPClassifier`. Start with a single hidden layer. You may use reasonable values for the other hyperparameters such as activation function, learning rate, etc. Report the performance of the model using appropriate metrics.

```python
# TODO: Create and evaluate the MLPClassifier pipeline
from sklearn.neural_network import MLPClassifier
pipe_mlp = make_pipeline(
    StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(100,), max_iter=500, random_state=27)
)
pipe_mlp.fit(X_train, y_train)
accuracy_mlp = pipe_mlp.score(X_test, y_test)
print(f"Test Accuracy (MLP): {accuracy_mlp:.2f}")

precision_mlp = precision_score(y_test, pipe_mlp.predict(X_test), average='weighted')
print(f"Test Precision (MLP): {precision_mlp:.2f}")
recall_mlp = recall_score(y_test, pipe_mlp.predict(X_test), average='weighted')
print(f"Test Recall (MLP): {recall_mlp:.2f}")
```


### 1.4 Hyperparameter tuning with `GridSearchCV`
Perform a small grid search using `GridSearchCV` on the following hyperparameter:
- number of hidden layers and size of each hidden layer
- `learning_rate_init`
- activation function
- batch size

Then retrain your model on the best parameters and report the performance of the model using appropriate metrics.

```python
# TODO: Grid search for MLPClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import f1_score
parameters = {
    'mlpclassifier__hidden_layer_sizes': [(50,), (100,), (150,),(50,50), (100,50), (150,50)],
    'mlpclassifier__learning_rate_init': [0.001, 0.01, 0.1],
    'mlpclassifier__activation': ['relu', 'tanh', 'logistic'],
    'mlpclassifier__batch_size': [16, 32, 64]
}
scoring_metrics = ['accuracy', 'precision_weighted', 'recall_weighted', 'f1_weighted']
grid_search = GridSearchCV(pipe_mlp, parameters, cv=5, scoring=scoring_metrics, refit='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)
print("Best parameters found: ", grid_search.best_params_)
print("Best Cross-Validation Accuracy Score:", grid_search.best_score_)

best_index = grid_search.best_index_
for metric in scoring_metrics:
    mean_score = grid_search.cv_results_['mean_test_' + metric][best_index]
    print(f"Best Cross-Validation {metric.replace('_', ' ').title()} Score: {mean_score:.4f}")
```
